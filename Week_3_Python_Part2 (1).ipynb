## Q3.5  Build an Invoice Line from Raw Values

```python
raw_name = ' wireless MOUSE '
raw_sku = 'sku-0017'
raw_price = '₹1,299.50'
raw_qty = ' 3 '
gst_rate = 18

# Clean the values
name = ' '.join(raw_name.split()).title()                  
sku = raw_sku.strip().upper()                              
price = float(raw_price.replace('₹', '').replace(',', '').strip())
qty = int(raw_qty.strip())

# Calculate the amounts
subtotal = round(price * qty, 2)
gst = round(subtotal * gst_rate / 100, 2)

total = round(subtotal + gst, 2)

# Cleaned values
for var, val in [('name', name), ('sku', sku), ('price', price), ('qty', qty),
                 ('subtotal', subtotal), ('gst', gst), ('total', total)]:
    print(f'{var:<9}{val!r}')

# Printed invoice line
print(f'{name} | {sku} | {qty} x {price:,.2f} = {subtotal:,.2f} | '
      f'GST {gst:,.2f} | Total {total:,.2f}')
```

## Q3.6  Update a Shopping Cart

```python
cart = {'P1': 2, 'P2': 1, 'P3': 5}

# Step 1: add P5 with quantity 1
cart['P5'] = 1


# Step 2: increase P2 by 3
cart['P2'] += 3

# Step 3: remove P3 and keep the removed quantity
removed_qty = cart.pop('P3')
print('cart after steps 1-3     :', cart)
print('quantity removed (step 3):', removed_qty)

# Step 4: P9 is not in the cart -> 0, no error
print('step 4                   :', cart.get('P9', 0))

# Step 5: product IDs in alphabetical order
print('step 5                   :', sorted(cart))

# Step 6: number of different products and total units
print('step 6                   :', f'{len(cart)} products, {sum(cart.values())} units')

# Step 7: product with the highest quantity
print('step 7                   :', max(cart, key=cart.get))
```

## Q3.7  Compare Customer IDs from Two Systems

```python
crm_ids = ['C01', 'C02', 'C03', 'C02', 'C04', 'C07']
billing_ids = ['C02', 'C04', 'C05', 'C06', 'C04']

crm_set, bill_set = set(crm_ids), set(billing_ids)

# A: CRM IDs without duplicates, first-appearance order
a = list(dict.fromkeys(crm_ids))
# B: in both systems (sorted)
b = sorted(crm_set & bill_set)
# C: only in CRM (sorted)
c = sorted(crm_set - bill_set)
# D: only in billing (sorted)
d = sorted(bill_set - crm_set)
# E: number of different IDs across both systems
e = len(crm_set | bill_set)

print('A', a)
print('B', b)

print('C', c)
print('D', d)
print('E', e)
```
